### Introduction to Bank Marketing

Introduction to Bank Marketing Dataset
The Bank Marketing dataset is obtained from the UCI Machine Learning Repository. It contains information about a Portuguese bank's direct marketing campaigns, which were conducted mainly through telephone calls to promote term-deposit subscriptions.
The dataset includes customer details such as age, job, marital status, education, account balance, and loan information. It also contains campaign details, including contact type, campaign duration, and the number of contacts made to each customer.
The main objective of this dataset is to analyze customer behavior and identify the factors that influence whether a customer subscribes to a term deposit. Data analysis can help banks improve their marketing strategies, target potential customers, and make better business decisions.

```python
import pandas as pd

# 1. Load the dataset
df = pd.read_csv("Bank marketing dataset.csv", sep=";")
```

```python

# 2. Check missing values
print(df.isnull().sum())
```

```python

# 3. Handle missing values
# Replace unknown values with NaN

df.replace(["ERROR", "UNKNOWN", ""], np.nan, inplace=True)
```

```python
# 4.Checking the number of duplicate rows
print("Duplicate Rows Found:",df.duplicated().sum())
```

```python
# 5.Checking Data Types of Each Column Before Conversion 
df.dtypes
```

```python
#Displaying Columns Before Derived
print("Current column names:",df.columns.tolist())
```

```python

# 6. Create derived columns
# Age group
df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 25, 35, 50, 100],
    labels=["Under 26", "26-35", "36-50", "Above 50"]
)
print(df[["age","age_group"]].head())
```

```python
# Convert subscription outcome into numeric format
df["subscribed"] = df["y"].map(
    {"yes": 1, "no": 0}
)
print(df[["y","subscribed"]].head())
```

```python
# Display the new columns
print(df[["age", "age_group", "y", "subscribed"]].head())
```

```python
# 7. Filtering data

# Select customers aged 30 and above
filtered_df = df[df["age"] >= 30]

# Display the first 5 rows
print(filtered_df.head())

```

```python
# Filter customers who subscribed to a term deposit.
subscribed_df = df[df["y"] == "yes"]

print(subscribed_df.head())
print("Total subscribed customers:", len(subscribed_df))
```

```python
# Filter customers with balance above 1000
high_balance = df[df["balance"] > 1000]

print(high_balance.head())
print("Total customers:", len(high_balance))
```

```python
# Filter customers with housing loans
housing_loan = df[df["housing"] == "yes"]

print(housing_loan.head())
print("Total customers with housing loans:", len(housing_loan))
```

### (ii) Univariate Analysis
It examines each feature individually to understand its basic distribution and key statistical properties.
Univariate analysis studies one variable at a time.


##### A. Distribution of customer age

```python
sns.histplot(df["age"], bins=20)
plt.title("Distribution of Customer Age")
plt.show()
```

Explanation of the graph

X-axis (Age): Shows the ages of bank customers.

Y-axis (Number of Customers): Shows how many customers belong to each age range.

Bars: Represent the number of customers in different age groups.

Insight – Distribution of Customer Age

The age distribution histogram helps to understand the demographic characteristics of bank customers. The dataset generally contains more customers in the middle-age groups, while relatively fewer customers are found at the youngest and oldest ages. This analysis helps the bank identify the age groups that make up its customer base. 

(B) Customer job categories

```python
df["job"].value_counts().plot(kind="bar")
plt.title("Customers by Job")
plt.xticks(rotation=45)
plt.show()
```

Insight – Customers by Job

The bar chart shows the distribution of bank customers across different job categories. Management has the highest number of customers, followed by blue-collar workers and technicians. Student and unknown job categories have comparatively fewer customers. This analysis helps the bank understand the occupational profile of its customers and identify the major job groups in its customer base.

(C) Term deposit subscription

```python
sns.countplot(data=df, x="y")
plt.title("Term Deposit Subscription")
plt.show()

print(df["y"].value_counts())
print(df["y"].value_counts(normalize=True) * 100)
```

Insight – Term Deposit Subscription

The count plot shows the distribution of customers according to their term deposit subscription status. Most customers did not subscribe, while a smaller number subscribed to the term deposit. This indicates an imbalance between the two categories, with non-subscribers forming the majority. The bank can use this information to understand its campaign outcomes and explore ways to improve customer conversion rates.

#### (iii) Bivariate Analysis
It examines the relationship between two variables to identify patterns,associations,and differences.It helps understand how one feature may influence or vary with another.

(A) Age and balance

```python
sns.scatterplot(data=df, x="age", y="balance")
plt.title("Age vs Account Balance")
plt.show()
```

Insights

Customers of different ages have different account balances.
Customers of the same age can have very different balances.
Some customers have high balances, while others have low or negative balances.
The relationship between age and account balance may not be strong or consistent.

```python
print(df["age"].corr(df["balance"]))
```

(B) Job and subscription

```python
sns.countplot(data=df, x="job", hue="y")
plt.title("Job vs Subscription")
plt.xticks(rotation=45)
plt.show()
```

Key insights from graph

Management: This job category has a high number of customers and also a high number of subscribers.

Blue-collar: This category has one of the highest numbers of customers who did not subscribe.

Technicians: A considerable number of customers did not subscribe, while fewer subscribed.

Retired customers: This group shows a noticeable number of subscriptions compared with several smaller job categories.

Students and unknown jobs: These categories have relatively fewer customers overall.

Overall observation: In every job category shown, the blue bar is taller than the orange bar. This means more customers did not subscribe than subscribed.


(C) Average balance by subscription


```python
print(df.groupby("y")["balance"].mean())
```


This compares the average account balance of customers who subscribed with those who did not.

### (iv) Groupby Analysis
Use groupby() to summarize customer information by category.

```python
# Average age by job
print(df.groupby("job")["age"].mean())

```

```python
# Subscription count by education
print(df.groupby(["education", "y"]).size())
```

```python

# Average balance by marital status
print(df.groupby("marital")["balance"].mean())
```

### (v) Pivot table analysis

A pivot table summarizes data across two categorical variables.

```python
pivot = pd.pivot_table(
    df,
    index="job",
    columns="y",
    values="balance",
    aggfunc="mean"
)

print(pivot)
```

This shows the average account balance for each job category, separated by subscription status.

### (vi) Correlation analysis

Correlation helps identify relationships between numerical variables.

```python
numeric = df.select_dtypes(include="number")

print(numeric.corr().round(2))

sns.heatmap(numeric.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

A positive correlation means two variables tend to increase together; a negative correlation means one tends to decrease as the other increases. Correlation does not prove causation.

select_dtypes(include="number") selects only numerical columns.

numeric.corr() calculates the correlation between numerical columns.

.round(2) rounds the results to two decimal places.

annot=True displays correlation values inside the heatmap.

cmap="coolwarm" uses colours to distinguish positive and negative correlations.

### (vii) Multivariate analysis

Multivariate analysis studies three or more variables together.

```python
sns.barplot(
    data=df,
    x="marital",
    y="balance",
    hue="y",
    errorbar=None
)

plt.title("Marital Status, Balance and Subscription")
plt.show()
```

This compares average balances by marital status and subscription outcome.

Insights 

Married customers: Compare the average balance of married customers who subscribed and those who did not.

Single customers: Compare the average balances of single customers in both subscription groups.

Divorced customers: Compare the average balances of divorced customers who subscribed and those who did not.

Overall observation: The graph helps identify differences in average account balance across marital status and subscription groups.
